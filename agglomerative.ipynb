```python
import os
import cv2
import numpy as np
from tqdm import tqdm
from sklearn.cluster import AgglomerativeClustering
from sklearn.decomposition import PCA
import matplotlib.pyplot as plt
```

```python
def load_images(folder_paths, target_size=(150, 150)):
    image_list = []

    for folder_path in folder_paths:
        file_list = [filename for filename in os.listdir(folder_path) if filename.endswith((".jpeg", ".jpg"))]

        for filename in tqdm(file_list, desc=f"Loading Images from {folder_path}", unit="image"):
            image_path = os.path.join(folder_path, filename)

            try:
                img = cv2.imread(image_path, cv2.IMREAD_GRAYSCALE)

                # Check if the image is None (corrupted or not found)
                if img is None:
                    print(f"Skipping {filename} - Unable to read the image.")
                    continue

                # Resize the image to the target size
                img = cv2.resize(img, target_size)

                img_array = np.array(img).reshape(-1)
                image_list.append(img_array)

            except Exception as e:
                print(f"Skipping {filename} - Error: {e}")
                continue

    return np.array(image_list)
```

```python
def plot_clusters(images, labels, n_clusters):
    pca = PCA(n_components=2)
    reduced_data = pca.fit_transform(images)

    plt.figure(figsize=(10, 6))
    for i in range(n_clusters):
        cluster_points = reduced_data[labels == i]
        plt.scatter(cluster_points[:, 0], cluster_points[:, 1], label=f'Cluster {i + 1}')

    plt.title('Agglomerative Clustering of Images')
    plt.xlabel('Principal Component 1')
    plt.ylabel('Principal Component 2')
    plt.legend()
    plt.show()
```

```python
folder_paths = ['data/lung_train/NORMAL', 'data/lung_train/ABNORMAL']
images = load_images(folder_paths)
```

```python
num_images = len(images)
print(f"Number of images: {num_images}")
```

```python
n_clusters = 2
clustering = AgglomerativeClustering(n_clusters=n_clusters, linkage='ward')
labels = clustering.fit_predict(images)
```

```python
labels
```

```python
plot_clusters(images, labels, n_clusters)
```

```python
test_folder_paths = ['data/lung_test/NORMAL', 'data/lung_test/ABNORMAL']
test_images = load_images(test_folder_paths)
```

```python
def predict_labels(model, new_images):
    # Use the trained model to predict labels for new images
    new_labels = model.fit_predict(new_images)
    return new_labels
```

```python
# Predict labels for new images
new_labels = predict_labels(clustering, test_images)

print("Predicted cluster labels for new images:", new_labels)
```

```python
def check_labels(labels, expected_first=1, expected_last=0):
    # Check if the first 500 images have the expected label
    first_group_labels = labels[:500]
    correct_first = np.sum(first_group_labels == expected_first)
    accuracy_first = correct_first / len(first_group_labels)

    # Check if the last 1000 images have the expected label
    last_group_labels = labels[-1000:]
    correct_last = np.sum(last_group_labels == expected_last)
    accuracy_last = correct_last / len(last_group_labels)

    print(f"Accuracy for the first 500 images: {accuracy_first * 100:.2f}%")
    print(f"Accuracy for the last 1000 images: {accuracy_last * 100:.2f}%")

    return accuracy_first, accuracy_last
```

```python
# Check and print accuracy for each group
accuracy_first, accuracy_last = check_labels(new_labels)
```

```python
# Calculate overall accuracy
overall_accuracy = (accuracy_first * 500 + accuracy_last * 1000) / (500 + 1000)
print(f"Overall Accuracy: {overall_accuracy * 100:.2f}%")
```
